```python
import torch
from transformers import GPT2Tokenizer, GPT2LMHeadModel

# Set device to GPU if available, otherwise CPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Load the tokenizer and the model with eager attention implementation
model_id = "gpt2"
tokenizer = GPT2Tokenizer.from_pretrained(model_id)
model = GPT2LMHeadModel.from_pretrained(model_id, attn_implementation="eager").to(device)

print("Model and tokenizer loaded successfully!")
```

```python
# 1. Define your prompt
prompt = "The future of artificial intelligence in analyzing visual data is"

# 2. Tokenize the input
inputs = tokenizer(prompt, return_tensors="pt").to(device)

# 3. Generate the output (Updated with advanced parameters)
outputs = model.generate(
    **inputs,                   # Unpacks both input_ids AND the attention_mask to fix the warning
    max_new_tokens=40,          # Generates exactly 40 NEW tokens after your prompt
    temperature=0.7,
    top_k=50,
    do_sample=True,
    repetition_penalty=1.2,     # Penalizes words the model has already used to prevent looping
    pad_token_id=tokenizer.eos_token_id
)

# 4. Decode the output
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("--- Generated Text ---")
print(generated_text)
```

```python
# 1. Import necessary libraries
import torch
from transformers import GPT2Tokenizer, GPT2LMHeadModel
from bertviz import head_view

device = "cuda" if torch.cuda.is_available() else "cpu"

# 2. THE FIX: Reload the model explicitly requesting "eager" attention
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2", attn_implementation="eager").to(device)

# 3. Tell the model to output its internal attention weights
model.config.output_attentions = True

# 4. Define your analytical prompt
text = "The artificial intelligence analyzed the visual data"
inputs = tokenizer(text, return_tensors="pt").to(device)

# 5. Run the forward pass (analyzing, not generating)
with torch.no_grad():
    outputs = model(**inputs)

# 6. Extract the attention data and tokens
attention = outputs.attentions
tokens = tokenizer.convert_ids_to_tokens(inputs.input_ids[0])

# 7. Render the interactive visualization (Must be the last line in the cell)
head_view(attention, tokens)
```
